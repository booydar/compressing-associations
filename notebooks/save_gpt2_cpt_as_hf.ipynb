```python
from transformers import GPT2LMHeadModel, AutoConfig
from safetensors.torch import load_file

# Load config and create model
config = AutoConfig.from_pretrained('/workspace-SR006.nfs2/bulatov/rmt/test-time/test_time_gd/runs/squad/gpt2_mem8_K2_mem_proj_whead_bs_64_lr_1e-04/run_1/checkpoint-13500')
model = GPT2LMHeadModel(config)

# Load weights from safetensors (strict=False because lm_head.weight is tied)
state_dict = load_file('/workspace-SR006.nfs2/bulatov/rmt/test-time/test_time_gd/runs/squad/gpt2_mem8_K2_mem_proj_whead_bs_64_lr_1e-04/run_1/checkpoint-13500/model.safetensors')
model.load_state_dict(state_dict, strict=False)

# Tie weights properly (lm_head shares weights with transformer.wte)
model.tie_weights()

# Save with save_pretrained
model.save_pretrained('/workspace-SR006.nfs2/bulatov/rmt/models/gpt2_checkpoint_13500')
```
